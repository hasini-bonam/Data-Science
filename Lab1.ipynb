```python
#1D Array Indexing
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print(arr[0]) 
print(arr[2]) 
print(arr[-1]) 
```

```python
#2D Array Indexing
import numpy as np
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]])
print(arr[0, 0])  
print(arr[1, 2])  
print(arr[2, 1])  
```

```python
#Negative indexing
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print(arr[-1]) 
print(arr[-2]) 
print(arr[-3]) 
```

```python
#Array slicing
import numpy as np
arr = np.array([10, 20, 30, 40, 50, 60])
print(arr[1:4])
```

```python
#Array Shape
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]])
print(arr.shape)
```

```python
#Array Reshaping
import numpy as np
arr = np.array([1, 2, 3, 4, 5, 6])
new_arr = arr.reshape(2, 3)
print(new_arr)
```

```python
#Looping 1D Array
for i in range(len(arr)):
    print("Index:", i, "Value:", arr[i])
```

```python
#Looping MultiD Array
import numpy as np
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]])
for row in arr:
    print(row)
```

```python
#Joining arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

result = np.concatenate((arr1, arr2))

print(result)
```
