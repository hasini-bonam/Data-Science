```python
import pandas as pd
from scipy.stats import spearmanr

# Example dataset
df = pd.DataFrame({
    'DS': [10, 20, 30, 40, 50],
    'CD': [12, 24, 33, 45, 60],
    'ASE': [15, 25, 35, 42, 55]
})

# Spearman correlation matrix
corr_matrix = df.corr(method='spearman')

print("Spearman Correlation Matrix:")
print(corr_matrix)

# Correlation between DS and CD
corr_DS_CD, p_DS_CD = spearmanr(df['DS'], df['CD'])
print("\nDS vs CD:")
print("Correlation:", corr_DS_CD)
print("P-value:", p_DS_CD)

# Correlation between DS and ASE
corr_DS_ASE, p_DS_ASE = spearmanr(df['DS'], df['ASE'])
print("\nDS vs ASE:")
print("Correlation:", corr_DS_ASE)
print("P-value:", p_DS_ASE)

# Correlation between CD and ASE
corr_CD_ASE, p_CD_ASE = spearmanr(df['CD'], df['ASE'])
print("\nCD vs ASE:")
print("Correlation:", corr_CD_ASE)
print("P-value:", p_CD_ASE)
```

```python
def hamming_distance(str1, str2):
    #ensure string are of equal length
    if len(str1) != len(str2):
        raise ValueError("Strings must be of equal length")
    #count differing positions
    return sum(ch1 != ch2 for ch1, ch2 in zip(str1, str2))
    #example usage
s1 = "karolin"
s2 = "kathrin"

dist = hamming_distance(s1, s2)

print(f"Hamming Distance between '{s1}' and '{s2}': {dist}")
```

```python
def jaccard_index(str1, str2):
    set1 = set(str1.split())
    set2 = set(str2.split())

    intersection = set1.intersection(set2)
    union = set1.union(set2)

    return len(intersection) / len(union)


# Example usage
s1 = "data science is fun"
s2 = "science makes data useful"

print("Jaccard Index:", jaccard_index(s1, s2))
```

```python
def lcs_length(x, y):
    m, n = len(x), len(y)

    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m):
        for j in range(n):
            if x[i] == y[j]:
                dp[i + 1][j + 1] = dp[i][j] + 1
            else:
                dp[i + 1][j + 1] = max(dp[i][j + 1], dp[i + 1][j])

    return dp[m][n]


seq1 = "ABCDEF"
seq2 = "AEBDF"

length = lcs_length(seq1, seq2)

print(f"Longest Common Subsequence length between '{seq1}' and '{seq2}': {length}")
```
