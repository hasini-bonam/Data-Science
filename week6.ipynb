```python
import pandas as pd
from sklearn.datasets import load_iris
from statistics import mode 
iris=load_iris()
df=pd.DataFrame(iris.data,columns=iris.feature_names)
print("First 5 rows of dataset:")
print(df.head())
mean_values=df.mean()
print("\nMean values:")
print(mean_values)
median_values=df.median()
print("\nMedian values:")
print(median_values)
mode_values =df.mode().iloc[0]
print("\nMode values:")
print(mode_values)
```

```python
import pandas as pd
import seaborn as sns
from statistics import mode
tips = sns.load_dataset("tips")
df = pd.DataFrame(tips)
print("First 5 rows of dataset:")
print(df.head())
mean_values = df.select_dtypes(include='number').mean()
print("\nMean values:")
print(mean_values)
median_values = df.select_dtypes(include='number').median()
print("\nMedian values:")
print(median_values)
mode_values = df.select_dtypes(include='number').mode().iloc[0]
print("\nMode values:")
print(mode_values)
```

```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

print("First 5 rows of dataset:")
print(df.head())

# Range (Max - Min)
range_values = df.max() - df.min()
print("\nRange (Max - Min):")
print(range_values)

# Variance
variance_values = df.var()
print("\nVariance:")
print(variance_values)

# Standard Deviation
std_values = df.std()
print("\nStandard Deviation:")
print(std_values)

# Interquartile Range (IQR)
iqr_values = df.quantile(0.75) - df.quantile(0.25)
print("\nInterquartile Range (IQR):")
print(iqr_values)
```

```python
import pandas as pd
import seaborn as sns

tips = sns.load_dataset("tips")
df = pd.DataFrame(tips)

print("First 5 rows of dataset:")
print(df.head())

# Select numerical columns
numeric_df = df.select_dtypes(include='number')

# Range (Max - Min)
range_values = numeric_df.max() - numeric_df.min()
print("\nRange (Max - Min):")
print(range_values)

# Variance
variance_values = numeric_df.var()
print("\nVariance:")
print(variance_values)

# Standard Deviation
std_values = numeric_df.std()
print("\nStandard Deviation:")
print(std_values)

# Interquartile Range (IQR)
iqr_values = numeric_df.quantile(0.75) - numeric_df.quantile(0.25)
print("\nInterquartile Range (IQR):")
print(iqr_values)
```
